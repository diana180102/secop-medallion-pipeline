```python
import psycopg
from pathlib import Path
from sqlalchemy import text, engine
import os
```

```python
DSN = os.getenv("DATABASE_URL")
RAIZ = Path.cwd().parent
```

```python
def run_sql_file(dsn:str, path:str) -> None:
    sql = Path(path).read_text(encoding="utf-8")
    with psycopg.connect(dsn) as conn:
        conn.execute(sql)
    print(f"ok: {path}")

run_sql_file(DSN, RAIZ / "sql" / "00_parseo_functions.sql")
run_sql_file(DSN, RAIZ / "sql" / "silver" / "01_silver_contratos_electronicos.sql")
```
