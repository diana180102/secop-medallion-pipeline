```python
import pandas as pd
from sqlalchemy import create_engine, types, text
import os
```

```python
#conexion a base de datos
db = os.getenv("DATABASE_URL")
engine = create_engine(db)
csv = 'data/contratos_electrónicos_20261013.csv'
name_table = 'contratos_electronicos'
schema = 'raw'
```

```python
# Celda NUEVA: crear el esquema
with engine.begin() as conn:
    conn.execute(text("CREATE SCHEMA IF NOT EXISTS raw"))
```

```python
#Estructura de la tabla
df_structure = pd.read_csv(csv, nrows=0)
types_sql = {col: types.Text() for col in df_structure.columns}
df_structure.to_sql(name_table, engine, schema=schema,  if_exists='replace', index=False)
```

```python
%%sql
```

```python
print("Date loading to docker container")
with pd.read_csv(csv, chunksize=100000, dtype=str) as reader:
    for i, chunk in enumerate(reader):
        chunk.to_sql(name_table, engine, schema=schema,  if_exists='append', index=False)
        print(f"Chunk {i + 1} successfully loaded")
```
